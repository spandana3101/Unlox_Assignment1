Q9.

1 / is normal division, so -7 / 2 = -3.5

2 // is floor division and rounds down, so -7 // 2 = -4

3 A remainder is indicated by %, and whose sign is the same as the sign of the divisor, thus: -7 % 2 = 1

4 * is power, so (-7) * 2 = 49

```python
#Q9.
a = -7
b = 2
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
```
