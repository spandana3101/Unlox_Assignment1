In matrix1 the comprehension builds three separate inner lists.In Matrix2,[[0]*3]*3 copies the reference to the same inner list three times, so all rows are one object and changing one row c hanges all of them

```python
#Q11.
matrix1 = [[0] * 3 for _ in range(3)]
matrix2 = [[0] * 3] * 3
matrix1[0][0] = 99
matrix2[0][0] = 99
print(matrix1)
print(matrix2)
```
