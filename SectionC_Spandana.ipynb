```python
#Q21.
income = float(input("Enter annual income (₹): "))

slabs = [(300000, 0), (700000, 5), (1000000, 10),
         (1200000, 15), (1500000, 20), (float("inf"), 30)]

lower = 0
total_tax = 0
for upper, rate in slabs:
    if income <= lower:
        break
    top = min(income, upper)
    tax = (top - lower) * rate / 100
    total_tax += tax
    print(f"₹{lower:>10,.0f} to ₹{top:>10,.0f} @ {rate:>2}%  ->  ₹{tax:>10,.2f}")
    lower = upper

print("-" * 50)
print(f"Total tax: ₹{total_tax:,.2f}")
```

```python
#Q31.
armstrong = []
for num in range(1, 10000):
    digits = 0
    temp = num
    while temp > 0:
        digits += 1
        temp //= 10
    total = 0
    temp = num
    while temp > 0:
        total += (temp % 10) ** digits
        temp //= 10
    if total == num:
        armstrong.append(num)

result = ""
for i in range(len(armstrong)):
    if i > 0:
        result += ", "
    result += str(armstrong[i])
print(result)
print("Count:", len(armstrong))
```

```python
#Q33.
def is_prime(num):
    if num < 2:
        return False
    i = 2
    while i * i <= num:
        if num % i == 0:
            return False
        i += 1
    return True

n = int(input("Enter N: "))
print(n, "is prime" if is_prime(n) else "is not prime")

prime_sum = 0
for k in range(2, n + 1):
    if is_prime(k):
        prime_sum += k
print(f"Sum of primes from 2 to {n}: {prime_sum}")
```

```python
#Q37.
marks = list(map(int, input("Enter 10 marks (space separated): ").split()))

highest = lowest = marks[0]
total = 0
first = second = -1          # first = highest, second = second highest (distinct)
for m in marks:
    total += m
    if m > highest:
        highest = m
    if m < lowest:
        lowest = m
    if m > first:
        second = first
        first = m
    elif m > second and m != first:
        second = m

average = total / len(marks)
above = 0
for m in marks:
    if m > average:
        above += 1

print("Maximum        :", highest)
print("Minimum        :", lowest)
print("Average        :", average)
print("Above average  :", above)
print("Second highest :", second)
```

```python
#Q38.
review = input("Enter a review: ")

cleaned = ""
for ch in review.lower():
    if ch.isalnum() or ch == " ":
        cleaned += ch

freq = {}
for w in cleaned.split():
    freq[w] = freq.get(w, 0) + 1

items = list(freq.items())
# bubble sort by count, highest first (stable, so ties keep their original order)
n = len(items)
for i in range(n):
    for j in range(n - 1 - i):
        if items[j][1] < items[j + 1][1]:
            items[j], items[j + 1] = items[j + 1], items[j]

for word, count in items:
    print(f"{word:<10} {count}")
if items:
    print("Most frequent:", items[0][0], "->", items[0][1])
```

```python
#Q41.
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

max_i = min_i = 0
for i in range(len(prices)):
    if prices[i] > prices[max_i]:
        max_i = i
    if prices[i] < prices[min_i]:
        min_i = i

average = sum(prices) / len(prices)

best_gain = prices[1] - prices[0]
best_i = 1
volatility = 0
for i in range(1, len(prices)):
    change = prices[i] - prices[i - 1]
    if change > best_gain:
        best_gain = change
        best_i = i
    volatility += abs(change)

print(f"(a) Max price : ₹{prices[max_i]} on {days[max_i]}")
print(f"(b) Min price : ₹{prices[min_i]} on {days[min_i]}")
print(f"(c) Average   : ₹{average:.2f}")
print(f"(d) Biggest gain: ₹{best_gain} between {days[best_i - 1]} and {days[best_i]}")
print(f"(e) Volatility: ₹{volatility}")
```

```python
#Q46.
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        base = amount / (1 + gst_rate / 100)
        gst = amount - base
        total = amount
    else:
        base = amount
        gst = amount * gst_rate / 100
        total = amount + gst
    return (round(base, 2), round(gst, 2), round(total, 2))

for result in (calculate_gst(1000),
               calculate_gst(1180, gst_inclusive=True),
               calculate_gst(500, gst_rate=5)):
    print(f"({result[0]:.2f}, {result[1]:.2f}, {result[2]:.2f})")
```

```python
#Q47.
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:
        return 1 / power(base, -exp)
    return base * power(base, exp - 1)

print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
#Q48.
import numpy as np
identity = np.eye(5)
print("(a) Identity:\n", identity)
scaled = identity * 10
print("(b) Times 10:\n", scaled)
# (c) Transposing swaps rows and columns, i.e. element [i][j] moves to [j][i].
# In an identity matrix the only non-zero values are on the diagonal (i == j),
# and everything else is 0, so swapping positions changes nothing.
transposed = scaled.T
print("(c) Transpose:\n", transposed)
print("(d) Sum of diagonal:", np.trace(scaled))
print("(e) Sum of all elements:", scaled.sum())
```
